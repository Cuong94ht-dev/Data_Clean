```python
import fitz  # Cài đặt bằng lệnh: pip install PyMuPDF
import os
import glob

def redact_phuthai_invoice(input_pdf, output_pdf):
    # Mở file PDF
    doc = fitz.open(input_pdf)
    page = doc[0] # Mặc định xử lý trang 1

    # 1. Các từ khóa làm mỏ neo để xác định vùng cần che
    start_text = "Họ tên người mua hàng:"
    end_text = "Số tài khoản:"

    # 2. Tìm kiếm tọa độ của các từ khóa này trên trang
    start_rects = page.search_for(start_text)
    end_rects = page.search_for(end_text)

    if start_rects and end_rects:
        # Lấy mép trên của chữ đầu và mép dưới của chữ cuối
        # Trừ/cộng thêm 8 pixel để che bao phủ đẹp mắt hơn
        y0 = start_rects[0].y0 - 8
        y1 = end_rects[0].y1 + 8

        # Cố định lề trái và lề phải cho form A4 chuẩn
        x0 = 40   # Bắt đầu sát lề trái
        x1 = 555  # Kéo dài đến sát lề phải

        # 3. Tạo vùng chữ nhật cần che
        target_area = fitz.Rect(x0, y0, x1, y1)

        # 4. Áp dụng lớp che màu trắng (fill=(1, 1, 1))
        page.add_redact_annot(target_area, fill=(1, 1, 1))
        page.apply_redactions() # Xóa vĩnh viễn dữ liệu gốc bên dưới
        
        doc.save(output_pdf)
        print(f"✅ Đã che thành công: {os.path.basename(input_pdf)}")
    else:
        print(f"❌ Bỏ qua: Form không chuẩn, không tìm thấy từ khóa trong {os.path.basename(input_pdf)}")

    doc.close()

# ==========================================
# CẤU HÌNH THƯ MỤC VÀ CHẠY HÀNG LOẠT
# ==========================================

# Bạn đổi tên 2 thư mục này cho đúng với máy của bạn nhé
input_folder = r"C:\Users\User\OneDrive - CONG TY CO PHAN TAP DOAN PHU THAI\Desktop\New folder"
output_folder = r"C:\Users\User\OneDrive - CONG TY CO PHAN TAP DOAN PHU THAI\Desktop\New folder (2)"

# Tạo thư mục đầu ra nếu chưa có
if not os.path.exists(output_folder):
    os.makedirs(output_folder)

# Quét và xử lý toàn bộ file PDF trong thư mục
pdf_files = glob.glob(f"{input_folder}/*.pdf")

if not pdf_files:
    print(f"Không tìm thấy file PDF nào trong thư mục '{input_folder}'")
else:
    print(f"Bắt đầu xử lý {len(pdf_files)} file...")
    for pdf_file in pdf_files:
        filename = os.path.basename(pdf_file)
        output_path = os.path.join(output_folder, filename)
        redact_phuthai_invoice(pdf_file, output_path)
        
    print("\n🎉 HOÀN THÀNH!")
```
