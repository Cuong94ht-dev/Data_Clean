```python
import os
import re
import shutil

def phan_loai_hoa_don(thu_muc_chua_pdf, thu_muc_chua_folder):
    """
    thu_muc_chua_pdf: Đường dẫn đến thư mục chứa các file PDF hóa đơn
    thu_muc_chua_folder: Đường dẫn đến thư mục chứa các folder chứng từ (409_, INV_...)
    """
    # 1. Lấy danh sách tất cả các folder hiện có
    cac_folder = [f for f in os.listdir(thu_muc_chua_folder) if os.path.isdir(os.path.join(thu_muc_chua_folder, f))]
    
    # 2. Lấy danh sách tất cả các file PDF
    cac_file_pdf = [f for f in os.listdir(thu_muc_chua_pdf) if f.lower().endswith('.pdf')]
    
    da_di_chuyen = 0
    khong_tim_thay = 0

    print("=== BẮT ĐẦU DI CHUYỂN FILE PDF HÓA ĐƠN ===")

    for file_name in cac_file_pdf:
        # Tách lấy tên file bỏ đuôi .pdf (ví dụ: 3700232756_1C26TPT_441)
        ten_file_goc, _ = os.path.splitext(file_name)
        
        # Lấy số HĐ sau dấu gạch dưới (_) cuối cùng
        parts = ten_file_goc.split('_')
        if not parts:
            continue
        so_hd = parts[-1].strip()  # Ví dụ: "441", "458"
        
        folder_dich = None
        
        # 3. Quét danh sách folder để tìm vị trí khớp với số HĐ
        for folder_name in cac_folder:
            # Trường hợp folder gom nhiều HĐ: INV(412 458)_(HFS)...
            if folder_name.upper().startswith("INV("):
                # Dùng Regex lấy cụm số nằm bên trong dấu ngoặc đơn ()
                match = re.search(r'\((.*?)\)', folder_name)
                if match:
                    # Tách các số hóa đơn bên trong dấu ngoặc bằng khoảng trắng
                    danh_sach_so_inv = match.group(1).split()
                    if so_hd in danh_sach_so_inv:
                        folder_dich = folder_name
                        break
            
            # Trường hợp folder đơn HĐ: 409_(HFS)... hoặc 411_(HFS)...
            else:
                # Lấy cụm số đứng trước dấu gạch dưới thứ nhất
                so_folder_don = folder_name.split('_')[0].strip()
                if so_hd == so_folder_don:
                    folder_dich = folder_name
                    break
                    
        # 4. Thực hiện di chuyển file nếu tìm thấy folder đích tương ứng
        if folder_dich:
            duong_dan_goc = os.path.join(thu_muc_chua_pdf, file_name)
            duong_dan_dich = os.path.join(thu_muc_chua_folder, folder_dich, file_name)
            
            try:
                shutil.move(duong_dan_goc, duong_dan_dich)
                print(f"✓ [Thành công] {file_name} -> Folder: {folder_dich}")
                da_di_chuyen += 1
            except Exception as e:
                print(f"❌ [Lỗi] Không thể di chuyển file {file_name}: {e}")
        else:
            print(f"⚠️ [Bỏ qua] Không tìm thấy folder nào khớp cho HĐ số: {so_hd} (File: {file_name})")
            khong_tim_thay += 1

    print("=========================================")
    print(f"HOÀN THÀNH! Đã di chuyển thành công: {da_di_chuyen} file PDF.")
    if khong_tim_thay > 0:
        print(f"Lưu ý: Có {khong_tim_thay} file không tìm thấy folder phù hợp.")

# === CẤU HÌNH ĐƯỜNG DẪN CỦA BẠN TẠI ĐÂY ===
# (Thay thế đường dẫn bên dưới bằng đường dẫn thực tế trên máy tính của bạn)
# Nếu file PDF và các folder nằm chung một chỗ, bạn điền 2 đường dẫn giống nhau.

THU_MUC_PDF = r"C:\Users\User\OneDrive - CONG TY CO PHAN TAP DOAN PHU THAI\Desktop\DOC122-pages"     # Thư mục chứa các file PDF hóa đơn
THU_MUC_FOLDER = r"C:\Users\User\OneDrive - CONG TY CO PHAN TAP DOAN PHU THAI\Desktop\BCT KHÁC P&G\T6.2026\APS+Manpower T6" # Thư mục chứa các folder chứng từ

phan_loai_hoa_don(THU_MUC_PDF, THU_MUC_FOLDER)
```
