```python
import win32com.client
import time

# Bắt đầu bấm giờ
start_time = time.time()

try:
    # Móc nối thẳng vào ứng dụng Excel đang chạy trên Windows
    excel = win32com.client.GetObject(Class="Excel.Application")
except Exception as e:
    print("❌ Lỗi: Không tìm thấy cửa sổ Excel nào đang mở. Bạn hãy mở file Excel lên trước nhé!")
    excel = None

if excel:
    # --- TỐI ƯU TỐC ĐỘ (ẨN THAO TÁC MÀN HÌNH) ---
    excel.ScreenUpdating = False  # Đóng băng màn hình để Excel không bị giật lag khi nhảy số
    excel.DisplayAlerts = False   # Tắt thông báo popup phiền phức
    excel.EnableEvents = False    # Tắt các sự kiện ngầm

    # Lấy Workbook (file) đang active (đang mở và hiển thị trên màn hình)
    wb = excel.ActiveWorkbook
    ws_form = wb.Sheets("PRF")

    # Vòng lặp chạy từ STT 140 đến 144 (để cận trên là 145)
    for stt in range(189, 202):
        print(f"Đang xử lý STT: {stt}")
        
        # 1. Gán STT vào ô M1
        ws_form.Range("M1").Value = stt
        
        # 2. Ép Excel tính toán lại công thức (VLOOKUP/INDEX...) ngay lập tức
        excel.Calculate()
        
        # 3. Gửi lệnh in
        ws_form.PrintOut()
        
        # 4. Thời gian chờ siêu ngắn để lệnh kịp xếp hàng vào máy in
        time.sleep(0.2) 

    # --- BẮT BUỘC: TRẢ LẠI TRẠNG THÁI BÌNH THƯỜNG ---
    # Vì file đang mở sẵn, nếu không bật lại ScreenUpdating, Excel của bạn sẽ bị "đóng băng" không bấm được gì cả
    excel.ScreenUpdating = True
    excel.DisplayAlerts = True
    excel.EnableEvents = True

    end_time = time.time()
    print(f"✅ Hoàn thành! Tổng thời gian chỉ mất: {round(end_time - start_time, 2)} giây")
```
